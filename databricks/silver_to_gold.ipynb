```python
import sys
import os

sys.path.append(
    os.path.abspath(os.path.join(os.getcwd(), "..", "..", "package"))
)

from transformations.aggregations import (
    create_city_summary,
    create_device_summary,
    create_network_summary
)


SILVER_TABLE = "workspace.default.telecom_silver"

CITY_TABLE = "workspace.default.telecom_city_summary"
DEVICE_TABLE = "workspace.default.telecom_device_summary"
NETWORK_TABLE = "workspace.default.telecom_network_summary"


# Read Silver
df_silver = spark.table(SILVER_TABLE)

# Create Gold datasets
df_city = create_city_summary(df_silver)
df_device = create_device_summary(df_silver)
df_network = create_network_summary(df_silver)

# Save Gold
df_city.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(CITY_TABLE)

df_device.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(DEVICE_TABLE)

df_network.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(NETWORK_TABLE)

print("Silver → Gold completed successfully")
```
